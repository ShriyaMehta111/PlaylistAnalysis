# Key Performance Indicators (KPI) Analysis

## Objective

Calculate the key business metrics of the France Top 50 Playlist dataset. These KPIs summarize the dataset and provide the foundation for the Power BI dashboard.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/france_top50_cleaned.csv")
print("Dataset shape: ",df.shape)
df.head()
```

## KPI 1: Total Songs

### Objective

Calculate the total number of playlist entries available in the dataset.

```python
total_songs = len(df)
print("Total Songs: ",total_songs)
```

## KPI 2: Total Unique Artists

### Objective

Calculate the total number of unique artists represented in the France Top 50 Playlist dataset.

```python
total_artists = df["artist"].nunique()
print("Total Unique artists :",total_artists)
```

## KPI 3: Average Popularity

### Objective

Calculate the average popularity score of songs in the France Top 50 Playlist dataset.

```python
average_popularity = round(df["popularity"].mean(),2)
print("Average Popularity :",average_popularity)
```

## KPI 4: Explicit Song Percentage

### Objective

Calculate the percentage of explicit songs in the France Top 50 Playlist dataset.

```python
explicit_percentage = round(df["is_explicit"].sum()/len(df)*100,2)
print("Explicit Song Percentage:", explicit_percentage, "%")
```

## KPI 5: Average Song Duration

### Objective

Calculate the average duration of songs in the France Top 50 Playlist dataset.

```python
average_duration = round(df["duration_minutes"].mean(), 2)
print("Average Song Duration:", average_duration, "minutes")
```

## KPI 6: Album vs Single Distribution

### Objective

Calculate the percentage distribution of album tracks and single releases in the France Top 50 Playlist dataset.

```python
album_single_percentage = (
    df[df["album_type"].isin(["album", "single"])]["album_type"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(album_single_percentage)
```

```python
album_percentage = album_single_percentage["album"]
single_percentage = album_single_percentage["single"]

print("Album Percentage:", album_percentage, "%")
print("Single Percentage:", single_percentage, "%")
```

## KPI 7: Average Album Size

### Objective

Calculate the average number of tracks available in albums represented in the France Top 50 Playlist dataset.

```python
average_album_size = round(df["total_tracks"].mean(), 2)
print("Average Album Size:", average_album_size)
```

## KPI 8: Most Frequent Artist

### Objective

Identify the artist with the highest number of appearances in the France Top 50 Playlist dataset.

```python
artist_counts = df["artist"].value_counts()

most_frequent_artist = artist_counts.idxmax()
artist_appearances = artist_counts.max()

print("Most Frequent Artist:", most_frequent_artist)
print("Number of Appearances:", artist_appearances)
```

## KPI 9: Most Popular Song

### Objective

Identify the song with the highest popularity score in the France Top 50 Playlist dataset.

```python
top_song = df.loc[df["popularity"].idxmax()]

print("Song:", top_song["song"])
print("Artist:", top_song["artist"])
print("Popularity:", top_song["popularity"])
```

## KPI 10: Most Popular Artist

### Objective

Identify the artist with the highest average popularity score in the France Top 50 Playlist dataset.

```python
artist_popularity = (
    df.groupby("artist")["popularity"]
      .mean()
      .round(2)
)

most_popular_artist = artist_popularity.idxmax()
highest_avg_popularity = artist_popularity.max()

print("Most Popular Artist:", most_popular_artist)
print("Average Popularity:", highest_avg_popularity)
```

## KPI 11: Clean Content Dominance Ratio

### Objective

Calculate the percentage of clean (non-explicit) songs in the France Top 50 Playlist dataset.

```python
clean_content_ratio = round(
    (df["is_explicit"].eq(False).sum() / len(df)) * 100, 2)
print("Clean Content Dominance Ratio:", clean_content_ratio, "%")
```

## KPI 12: Content Acceptance Score

### Objective

Measure the overall audience acceptance of playlist content using the average popularity score.

```python
content_acceptance_score = round(df["popularity"].mean(), 2)

print("Content Acceptance Score:", content_acceptance_score)
```
