# Handling missing and invalid values

```python
df = df.dropna(subset = ["song"])
df.isnull().sum()
df.shape
```

# Duplicate record analysis

```python
duplicate_counts = df.duplicated().sum()
print(f"Total duplicates are : {duplicate_counts}")
```

```python
duplicates = df[df.duplicated(keep=False)]
duplicates
```

# Remove duplicate records

```python
df = df.drop_duplicates()
print("Duplicate rows :",df.duplicated().sum())
print(df.shape)
```

# Validate categorical columns

```python
df["album_type"].unique()
```

```python
df["is_explicit"].unique()
```

# Validate daily playlist size

```python
songs_per_day= df.groupby("date").size()
songs_per_day.value_counts()
```

```python
songs_per_day[songs_per_day!=50]
```

```python
march1 = df[df["date"]=="2025-03-01"]
march1["position"].value_counts().sort_index()
```

```python
march1[march1.duplicated(subset=["position"],keep=False)].sort_values("position")
```

```python
sorted(march1["position"].unique())
```

```python
march1 = march1.drop_duplicates(subset="position",keep="first")
other_days = df[df["date"]!="2025-03-01"]
df = pd.concat([other_days,march1], ignore_index=True)
df.sort_values(["date","position"]).reset_index(drop=True)
```

```python
songs_per_day = df.groupby("date").size()
songs_per_day.value_counts()
```

```python
feb2 = df[df["date"]=="2025-02-02"]
feb2["position"].sort_values().tolist()
```

# Convert date column

```python
df["date"] = pd.to_datetime(df["date"], format="%d-%m-%Y")
df["date"].dtype
```

```python
df["date"].head()
```

# Convert song duration to Minutes

```python
df["duration_minutes"] = df["duration_ms"]/60000
df["duration_minutes"] = df["duration_minutes"].round(2)
df[["duration_minutes","duration_ms"]].head()
```

```python
df.to_csv("../data/france_top50_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Data Cleaning Summary

### Cleaning Steps Performed

- Loaded the France Top 50 Playlist dataset.
- Inspected dataset structure using head(), info(), describe(), and shape.
- Identified one corrupted record with a missing song name, zero popularity, and zero duration. This record was removed.
- Detected and removed exact duplicate rows.
- Converted the "date" column to datetime format.
- Converted song duration from milliseconds to minutes.
- Validated daily playlist sizes.
- Investigated playlist anomalies:
  - One day contained 49 valid songs after removing a corrupted record.
  - One day contained multiple entries for the same playlist positions, indicating merged playlist snapshots. The first occurrence of each position was retained to restore the Top 50 playlist structure.
